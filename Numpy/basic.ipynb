## 1. Variables, objects and basic types

Python variables are names pointing to objects. The same variable name can later point to an object of a different type.

````python
name = "Alex"
age = 20
height = 1.75
is_student = True

print(name)
print(age)
print(height)
print(is_student)

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
````

## 2. Operators

````python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)       # true division
print(a // b)      # floor division
print(a % b)       # remainder
print(a ** b)      # power

print(a > b)
print(a == b)
print(a != b)
````

### Operator groups

- Arithmetic: `+ - * / // % **`
- Comparison: `== != < <= > >=`
- Logical: `and or not`
- Membership: `in`, `not in`
- Identity: `is`, `is not`

**Important:** `==` asks whether values are equal. `is` asks whether two names refer to the same object. Don't use `is` as a general replacement for `==`.

## 4. Lists — the most important prerequisite for NumPy

A list is an ordered, mutable Python collection.

```python
numbers = [10, 20, 30, 40]
```

Before learning NumPy indexing, you should be completely comfortable with list indexing and slicing.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])
print(numbers[-1])

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
print(numbers[::-1])
````

````python
numbers = [10, 20, 30]

numbers.append(40)
numbers.extend([50, 60])
numbers.insert(1, 15)

print(numbers)

numbers[0] = 999
print(numbers)
````

### List indexing mental model

For:

```text
[10, 20, 30, 40, 50]
  0   1   2   3   4
 -5  -4  -3  -2  -1
```

Remember:

```python
items[start:stop:step]
```

`stop` is **exclusive**.

## 5. Nested lists → the bridge to 2D arrays

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix)
print(matrix[0])
print(matrix[1][2])
print(matrix[2][0])
````

A nested Python list can represent a grid, but it is **not yet a NumPy matrix**.

Conceptually:

```text
matrix[row][column]
```

Later, NumPy lets you write the cleaner multidimensional form:

```python
array[row, column]
```

## 9. Conditions

````python
score = 78

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "D"

print(grade)
````

## 10. Loops

````python
numbers = [10, 20, 30, 40]

for number in numbers:
    print(number)

for index, number in enumerate(numbers):
    print(index, number)
````

````python
count = 0

while count < 5:
    print(count)
    count += 1
````

### `break`, `continue`, `pass`

- `break` → leave the loop.
- `continue` → skip the current iteration.
- `pass` → intentionally do nothing; useful as a placeholder.

````python
for number in range(10):
    if number == 3:
        continue
    if number == 7:
        break
    print(number)
````

## 11. Functions

````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
````

````python
def calculate_average(values):
    if not values:
        return None
    return sum(values) / len(values)

scores = [80, 90, 70, 100]
print(calculate_average(scores))
````

## 12. Scope

````python
x = "global"

def demo():
    x = "local"
    print(x)

demo()
print(x)
````

A local variable normally belongs to the function where it is created. Avoid unnecessary global state; it becomes especially confusing in notebooks.

## 13. Comprehensions

````python
squares = [x ** 2 for x in range(10)]
even_squares = [x ** 2 for x in range(10) if x % 2 == 0]

print(squares)
print(even_squares)
````

````python
names = ["alice", "BOB", "charlie"]
clean = [name.title() for name in names]

print(clean)
````

Comprehensions are useful, but don't force them when a normal loop is clearer.

The NumPy equivalent of many element-by-element loops is **vectorization**, which we will learn separately.

## 14. `zip`, `enumerate`, `sorted`, `sum`, `min`, `max`

````python
names = ["A", "B", "C"]
scores = [80, 90, 75]

for name, score in zip(names, scores):
    print(name, score)

print(list(enumerate(names)))
print(sorted(scores))
print(sum(scores))
print(min(scores))
print(max(scores))
````

## 15. Mutability, references and copies

````python
a = [1, 2, 3]
b = a

b.append(4)

print(a)
print(b)
print(a is b)
````

`b = a` does not create a new list. Both names refer to the same list.

For an independent shallow copy:

```python
b = a.copy()
```

This idea becomes important in NumPy because **views and copies** can behave differently.

````python
a = [1, 2, 3]
b = a.copy()

b.append(4)

print(a)
print(b)
print(a is b)
````

## 16. Errors and exceptions

````python
try:
    value = int("abc")
except ValueError as error:
    print("Could not convert:", error)
````

````python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero.")
finally:
    print("This runs either way.")
````

Use exceptions for situations that can genuinely fail. Don't hide every bug with `except Exception: pass`. That makes debugging harder.

## 19. Python list vs NumPy array — the key transition

````python
python_list = [1, 2, 3]
print(python_list * 2)  # repeats the list

# NumPy will instead interpret multiplication numerically:
# np.array([1, 2, 3]) * 2 -> [2, 4, 6]
````

This difference is one of the reasons NumPy exists.

A Python list is a general-purpose container.

A NumPy array is a numerical data structure designed for efficient operations over homogeneous data.
