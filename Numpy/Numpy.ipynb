## 17. Copy vs View

````python
# assignment does NOT create an independent array
a = np.array([1, 2, 3, 4])
b = a

b[0] = 100

print("a:", a)
print("b:", b)
````

````python
# copy creates independent data
a = np.array([1, 2, 3, 4])
b = a.copy()

b[0] = 100

print("a:", a)
print("b:", b)
````

````python
# slicing usually creates a view
a = np.array([1, 2, 3, 4])
b = a[1:3]

b[0] = 100

print("a:", a)
print("b:", b)
````

## 35. Broadcasting Rules

When NumPy compares shapes for broadcasting, it works from the **rightmost dimension**.

Examples:

```text
(2, 3)
(3,)
```

compatible:

```text
(2, 3)
(   3)
```

Result:

```text
(2, 3)
```

Another example:

```text
(2, 3)
(2, 1)
```

compatible and produces:

```text
(2, 3)
```

But:

```text
(2, 3)
(2,)
```

is not compatible.

The full rules become much easier once the shape foundation in `basic.ipynb` is clear.

## 39. Linear Algebra

````python
A = np.array([
    [1, 2],
    [3, 4]
])

# transpose
print(A.T)

# determinant
print("det:", np.linalg.det(A))

# inverse
print("inverse:")
print(np.linalg.inv(A))
````

````python
# eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("eigenvalues:", eigenvalues)
print("eigenvectors:")
print(eigenvectors)
````

````python
# solve Ax = b
A = np.array([
    [3, 1],
    [1, 2]
])

b = np.array([9, 8])

x = np.linalg.solve(A, b)

print(x)
````

## 46. Feature Scaling Example

````python
X = np.array([
    [10, 100],
    [20, 200],
    [30, 300],
    [40, 400]
], dtype=float)

mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

X_standardized = (X - mean) / std

print("mean:", mean)
print("std:", std)
print("standardized:")
print(X_standardized)
````
